# Data Cleaning

### Create a copy of the original dataset for cleaning while keeping the original data safe

```python
airbnb = df.copy()
```

```python
airbnb.columns
```

Rename the column `hourse_rules` to `house_rules` as there is a spelling mistake

```python
airbnb.columns
```

#### The column `img_links` has the image links and `id` is only a unique identifier.

```python
airbnb.drop(columns=["id","img_links"], inplace=True)
```

### Duplicate Records

Identify and removing of Duplicate records to ensure that each Airbnb is represented only once in the dataset.


```python
duplicates = airbnb.duplicated().sum()
print(f"Duplicate Rows: {duplicates}")
```

Drop the 599 duplicate records found above

## We need to convert the data type of `price`, `rating` and `reviews` into numeric for performing necessary analysis

*But before converting, we need to need to clean the values properly, so that we don't face value error while converting*

```python
airbnb["price"].unique()[:50]
```

#### Now, we would remove the symbols and unnecessary texts so that, we can convert the datatype to numeric to perform calculation and analysis as needed

```python
airbnb["price"]=(
    airbnb["price"]
    .str.replace("₹","")
    .str.replace(",","")
    .str.replace("per night","")
    .str.replace(" ","")
)
```

```python
airbnb["price"].unique()[:50]
```

```python
airbnb['price']=pd.to_numeric(airbnb['price'])
```

Converting the 'new' or 'New' string values in `rating` column to 'Nan' float values using numpy

```python
airbnb['rating']=airbnb['rating'].replace(['New','NEW'], np.nan)
```

Now, converting the data type of `rating` to numeric.

```python
airbnb['rating']=pd.to_numeric(airbnb['rating'])
```

```python
airbnb["reviews"].unique()[:50]
```

Converting the 'nan' or 'Nan' string values in `reviews` column to 'Nan' float values using numpy

```python
airbnb['reviews']=airbnb['reviews'].replace(['nan','Nan'], np.nan)
```

```python
airbnb['reviews']= (airbnb["reviews"]
                    .str.replace(",","")
)
```

```python
airbnb['reviews']=pd.to_numeric(airbnb['reviews'])
```

#### Check the columns with data type 'object' now

```python
text_columns = airbnb.select_dtypes(include='object').columns
print(text_columns)
```

Remove any extra unnecessary spaces from the data.

```python
for col in text_columns:
  airbnb[col]=airbnb[col].str.strip()
```

Check the number and percentage of missing data for each column.

Here, I would like to remove the column `host_id` first, as it is just extra data and would not be helping with any business questions

```python
airbnb.drop(columns="host_id",inplace=True)
```

```python
airbnb.columns
```

### Drop the records with null values in `name`, `host_name`, `address` and `features` columns as their percentage is very small and removing them would not not have any significant changes in our dataset analysis.

```python
airbnb.dropna(
    subset=["name","host_name","address","features"],
    inplace=True
    )
```

Filling the null values in `amenities`, `safety_rules` and `house_rules` with appropriate string values representing absence of those data.

```python
airbnb["amenities"]=airbnb["amenities"].fillna("No Amenities Listed")
airbnb["safety_rules"]=airbnb["safety_rules"].fillna("No Safety Rules Provided")
airbnb["house_rules"]=airbnb["house_rules"].fillna("No House Rules Provided")
```

Checking the status of null values now.

```python
airbnb.isnull().sum()
```

```python
airbnb.duplicated().sum()
```

Just in case, if there is any multiple spaces in the dataset. We would be removing that

```python
for col in text_columns:
  airbnb[col] = airbnb[col].str.replace(r"\s+"," ")
```

The `features` and `house_rules` columns contain multiple pieces of information in a single text field.

To improve the dataset for analysis, these fields are broken down into separate columns. This will enable more detailed exploratory analysis and business insights.

```python
airbnb['features'].head(10)
```

```python
airbnb['house_rules'].head(10)
```

```python
airbnb['guests']=(airbnb['features']
                  .str.extract(r'(\d+)\s*guest?')
                  .astype("Int64")
)
```

```python
airbnb['bedrooms']=(airbnb['features']
                    .str.extract(r'(\d+)\s*bedroom?')
                    .astype("Int64"))
```

```python
airbnb['beds']=(airbnb['features']
                .str.extract(r'(\d+)\s*bed?')
                .astype("Int64"))
```

```python
airbnb['bathrooms']=(airbnb['features']
                     .str.extract(r'(\d+)\s*bathroom?')
                     .astype("Int64"))
```

We extract the numbers and set the datatype of the columns as 'Int64'

*Note that its not 'int64' but 'Int64' as 'int64' is not capable of containing null values*

```python
airbnb[['features','guests','bedrooms','beds','bathrooms']].head()
```

```python
airbnb['check_in']=(airbnb['house_rules']
                    .str.extract(r'Check-in:\s*([^,]+)'))
```

```python
airbnb['check_out']=(airbnb['house_rules']
                     .str.extract(r'Check out:\s*([^,]+)'))
```

Similarly, we also extract `check_in` and `check_out` from the `house_rules` column.

Now, we extract the country from the `address` column and store it in a new column.

```python
airbnb['country']=(airbnb['address']
                   .str.split(",")
                   .str[-1]
                   .str.strip()
                   )
```

```python
airbnb['country'].head(10)
```

```python
airbnb[['house_rules','check_in','check_out']].sample(10)
```

Now, we remove the feature column as it no longer contains any data that is not extracted and hence is now not useful to us.

```python
airbnb.drop(columns='features',inplace=True)
```

Now, we remove the unnecessary symbols that is not needed and can probably interfere with the analysis.

```python
airbnb['safety_rules']=airbnb['safety_rules'].str.replace('\U000f1803',"",regex=False)
airbnb['safety_rules']=airbnb['safety_rules'].str.replace(',,',",",regex=False)
```

```python
row = airbnb.loc[0, "safety_rules"]

for i, ch in enumerate(row):
    print(i, repr(ch))
```

# Data Cleaning Summary
The raw Airbnb dataset was cleaned and now is ready for analysis.
The following steps were performed:

- Corrected misspelled column names.
- Removed irrelevant columns (`id`, `host_id` and `img_links`) and records with missing important information (`name`, `host_name`, `address`, and `features`).
- Filled missing values in text-based columns with descriptive statements.
- Retained missing values in `rating` and `reviews` as they represent unrated or newly listed properties.
- Removed duplicate records.
- Standardized text formatting by removing any unnecessary whitespaces.
- Extracted the number of guests, bedrooms, beds, and bathrooms from the `features` column.
- Extracted check-in and check-out timings from the `house_rules` column.
- Removed the original `features` column after extracting its information.
